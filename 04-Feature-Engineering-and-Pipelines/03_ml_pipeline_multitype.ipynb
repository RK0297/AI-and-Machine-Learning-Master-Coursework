## Pipelines for numerical and categorical data must be separate

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
import joblib
from sklearn.pipeline import make_pipeline, Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
```

```python
d2 = {'Genre':['Rock', 'Metal', 'Bluegrass', 'Rock', np.nan, 'Rock', 'Rock', np.nan, 'Bluegrass', 'Rock'],
    'Social_media_followers':[1000000, np.nan, 2000000, 1310000, 1700000, np.nan, 4100000, 1600000, 2200000, 1000000],
    'Sold_out':[1,0,0,1,0,0,0,1,0,1]}
```

```python
df = pd.DataFrame(data=d2)
df.head(10)
```

```python
X = df.iloc[:, 0:2]
y = df.iloc[:, 2]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=75)
```

```python
num_cols = ['Social_media_followers']
cat_cols = ['Genre']
```

```python
num_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='mean')),
    ('scale',StandardScaler())
])
```

## Numerical Pipeline

The **numerical pipeline** handles numeric data (like 'Social_media_followers'):

**Step 1: Impute (Fill Missing Values)**
- Strategy: `mean` - fills NaN with the average of the column
- Example: If a follower count is missing, replace it with the average follower count

**Step 2: Scale (Standardize)**
- `StandardScaler` transforms numbers to have mean=0 and std=1
- Makes all numeric features on the same scale (e.g., 1000 to 2000 range → 0 to 1 range)
- Why? Models train better when features have similar ranges; prevents large-valued features from dominating

```python
cat_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('one-hot',OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
```

## Categorical Pipeline

The **categorical pipeline** handles text/category data (like 'Genre'):

**Step 1: Impute (Fill Missing Values)**
- Strategy: `most_frequent` - fills NaN with the most common value
- Example: If a genre is missing, replace it with the most popular genre in the data
- Why different from numerical? Categories don't have a meaningful "average"

**Step 2: One-Hot Encode**
- Converts categories into binary columns (0 or 1)
- Example: Genre=['Rock', 'Metal', 'Bluegrass'] becomes 3 binary columns:
  - is_Rock: [1, 0, 0, 1, ...]
  - is_Metal: [0, 1, 0, 0, ...]
  - is_Bluegrass: [0, 0, 1, 0, ...]
- `handle_unknown='ignore'`: If a new genre appears, set all columns to 0
- `sparse_output=False`: Returns dense array (easier to work with)

```python
col_trans = ColumnTransformer(transformers=[
    ('num_pipeline',num_pipeline,num_cols),
    ('cat_pipeline',cat_pipeline,cat_cols)
    ],
    remainder='drop',
    n_jobs=-1)
```

## ColumnTransformer - Processing Different Data Types

When you have mixed data types (numerical AND categorical), you can't use the same preprocessing for both. **ColumnTransformer** solves this by:

1. **Applying different pipelines to different columns** based on their data type:
   - Numerical columns: Impute with mean → Scale/Normalize
   - Categorical columns: Impute with most frequent → One-Hot Encode

2. **Key parameters:**
   - `transformers`: List of (name, pipeline, column_names) tuples
   - `remainder='drop'`: Drop any columns not specified in transformers
   - `n_jobs=-1`: Use all CPU cores for parallel processing

3. **Why separate pipelines?**
   - Numbers need scaling to be on the same range
   - Categories need encoding to become numbers
   - Different imputation strategies (mean vs. most frequent)

```python
dtc = DecisionTreeClassifier()
```

```python
pipefinal = make_pipeline(col_trans, dtc)
```

```python
pipefinal.fit(X_train, y_train)
```

```python
pipefinal.score(X_test, y_test)
```

```python
joblib.dump(pipefinal,"pipe.joblib")
```

## Saving & Loading Models with joblib

Once you've trained a complex pipeline, you want to save it so you can use it later WITHOUT retraining.

**joblib.dump()** - Save the trained model to a file:
- Takes the trained pipeline object and a filename
- Saves all learned parameters (mean values, scaling factors, model coefficients, etc.)
- Creates a binary file that preserves the entire model state

**joblib.load()** - Load the saved model back:
- Reads the binary file and reconstructs the pipeline object
- You can immediately use it to make predictions on new data
- No need to retrain!

**Why joblib instead of pickle?**
- joblib is optimized for scikit-learn models
- Handles large numpy arrays efficiently
- Better performance with complex pipelines

```python
pipefinal2 = joblib.load("pipe.joblib")

```
