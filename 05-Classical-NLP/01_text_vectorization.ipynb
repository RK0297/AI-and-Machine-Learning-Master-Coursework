## Define Sample Text Data

Create a small corpus of sentences about AI and machine learning. This will be our dataset for demonstrating different vectorization techniques.

**Example:** Each sentence in the list represents a document that we want to convert into numerical format.

# Text Vectorization Techniques

This notebook demonstrates different methods to convert text into numerical representations (vectors) for machine learning applications.

```python
text_data = [
    "AI is transforming the world.",
    "Natural language processing is a subset of AI.",
    "Deep learning and machine learning are popular AI techniques.",
    "AI applications are diverse and growing rapidly."
]
```

## Tokenize the Text Data

Break down each sentence into individual words (tokens) and convert them to lowercase. Tokenization is the first step in text processing.

**What happens:**
- "AI is transforming the world." → `['ai', 'is', 'transforming', 'the', 'world', '.']`
- Punctuation is kept as separate tokens
- All text is converted to lowercase for consistency

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download("punkt_tab")

tokenized_data = [
    word_tokenize(sentence.lower())
    for sentence in text_data
]
print(tokenized_data)
```

## Create Vocabulary

Build a vocabulary (unique set of all words) from all sentences. This vocabulary represents all possible words in our corpus.

**Process:**
1. Flatten all tokens from all sentences into one list
2. Use `set()` to get unique words only
3. Count the total vocabulary size

**Example:** If "AI" appears in 3 sentences, it's only counted once in the vocabulary.

```python
flattened_tokens = [
    word
    for sentence in tokenized_data
    for word in sentence
]

vocabulary = list(set(flattened_tokens))
print("Vocabulary:", vocabulary)
print("Vocabulary Size:", len(vocabulary))
print("Flattened Tokens:", flattened_tokens)
```

## Method 1: Count Vectorization (Bag of Words)

Use sklearn's CountVectorizer to convert text into numerical vectors. Each sentence becomes a vector where each position represents a word from the vocabulary, and the value is how many times that word appears.

**Example:**
- Vocabulary: `['ai', 'is', 'transforming', 'world', ...]`
- Sentence: "AI is transforming the world"
- Vector: `[1, 1, 1, 1, ...]` (each word appears once)

**Note:** This is called "Bag of Words" because word order is ignored.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
vectorized_data = vectorizer.fit_transform(text_data)

feature_names = vectorizer.get_feature_names_out()
vector_array = vectorized_data.toarray()

feature_names, vector_array
```

## Method 2: One-Hot Encoding Setup

Initialize a One-Hot Encoder that will create binary vectors for each word. This is a different approach from count vectorization.

**Configuration:**
- `handle_unknown="ignore"`: Skip words not in vocabulary
- `sparse_output=False`: Return regular arrays (not sparse matrices)

The encoder is trained on our vocabulary to learn all possible words.

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(np.array(vocabulary).reshape(-1, 1))
```

## Apply One-Hot Encoding

Transform each sentence into a one-hot encoded vector by:
1. Encoding each word individually as a one-hot vector
2. Summing all word vectors to get a sentence vector

**Example:**
- Word "ai": `[1, 0, 0, 0, ...]` (1 at position for "ai")
- Word "is": `[0, 1, 0, 0, ...]` (1 at position for "is")
- Sentence: `[1, 1, 0, 0, ...]` (sum of all word vectors)

The result is similar to Bag of Words, but built using one-hot encoding.

```python
one_hot_encoded_data = []

for sentence in tokenized_data:
    encoded_sentence = encoder.transform(
        np.array(sentence).reshape(-1, 1)
    )
    one_hot_encoded_data.append(
        np.sum(encoded_sentence, axis=0)
    )

one_hot_encoded_data = np.array(one_hot_encoded_data)
```

## View Encoded Result

Display the one-hot encoded vector for the first sentence to see the numerical representation.

**Interpretation:** Each number represents how many times the corresponding vocabulary word appears in the first sentence ("AI is transforming the world.").

```python
one_hot_encoded_data[0]
```

## Method 3: TF-IDF Vectorization
TF-IDF (Term Frequency-Inverse Document Frequency) scales the count of a word by how rarely it appears across all documents. Frequent words like "is" or "the" get lower scores, while unique meaning-carrying words get higher scores.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_vectorized_data = tfidf_vectorizer.fit_transform(text_data)

tfidf_feature_names = tfidf_vectorizer.get_feature_names_out()
tfidf_array = tfidf_vectorized_data.toarray()

print("TF-IDF Features:", tfidf_feature_names)
print("TF-IDF Array Shape:", tfidf_array.shape)
# Look at the first sentence to see the scaled weights:
print("TF-IDF for sentence 1:", tfidf_array[0])
```
