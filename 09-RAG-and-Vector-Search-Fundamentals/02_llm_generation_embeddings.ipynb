```python
from transformers import AutoTokenizer

# Load a pre-trained tokenizer (e.g., BERT)
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Tokenize a sentence
text = "Hey there, my name is RK"
input_tokens = tokenizer(text, return_tensors="pt")

print("Tokenized Output (Input IDs):")
print(input_tokens["input_ids"])
```

```python
from openai import OpenAI
import os

# Set up OpenRouter API Key with base URL
# SECURITY: Never commit real API keys. Use environment variables.
api_key = os.environ.get("OPENROUTER_API_KEY", "sk-or-v1-YOUR_API_KEY_HERE")

client = OpenAI(
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1"
)

# Create vector embeddings for a text
response = client.embeddings.create(
    input="Cat loves milk",
    model="text-embedding-3-small"
)

# The result is a list of 1536 numbers (dimensions)
embeddings = response.data[0].embedding
print(f"Embedding Length: {len(embeddings)}")
print(f"Sample values: {embeddings[:5]}")
```

```python
import torch
from transformers import AutoModelForCausalLM

# Load the model with float16 precision for efficiency
model = AutoModelForCausalLM.from_pretrained(
    model_name, 
    torch_dtype=torch.bfloat16
)

# Generate a response based on a prompt
prompt = "Write a Python code for adding two numbers:\n"
input_ids = tokenizer(prompt, return_tensors="pt")

# Generate next tokens (predicting words) with attention mask and sampling
gen_out = model.generate(
    input_ids=input_ids['input_ids'],
    attention_mask=input_ids['attention_mask'],
    max_new_tokens=100,
    pad_token_id=tokenizer.eos_token_id,
    do_sample=True,
    temperature=0.7,
    top_p=0.9
)

# Decode tokens back into human-readable text
readable_output = tokenizer.batch_decode(gen_out, skip_special_tokens=True)
print(readable_output[0])
```
